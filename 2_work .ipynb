```python
import pandas as pd
import numpy as np

# Load the teaching dataset
df = pd.read_excel(“teaching_prices_messy (1).xlsx”)

# Clean column names
df.columns = df.columns.str.strip().str.lower()

# Convert prices into numbers
df[“price”] = pd.to_numeric(df[“price”], errors=“coerce”)

# Convert dates
df[“date”] = pd.to_datetime(df[“date”], errors=“coerce”)

# Remove rows without essential information
df = df.dropna(subset=[“date”, “item”, “price”])

# Display the cleaned data
print(df.head())
print(“\nTotal observations:”, len(df))
print(“\nItems:”, df[“item”].unique())
print(“\nCities:”, df[“city”].unique())
```
